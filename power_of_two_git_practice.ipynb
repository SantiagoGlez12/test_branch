```python
def es_potencia_de_2(numero):
    if isinstance(numero, bool) or not isinstance(numero, int) or numero < 1:
        return False
    while numero > 1:
        if numero % 2 != 0:
            return False
        numero //= 2
    return True
```

```python
numeros = [2, 4, 8, 10, 16, 32]
for numero in numeros:
    print(f"{numero}: {es_potencia_de_2(numero)}")
```
